```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px

# Load the dataset
df = pd.read_csv('improved_global_startups_dataset.csv')

# Display basic info
print("Shape of dataset:", df.shape)
print("\nColumn Types:\n", df.dtypes)
print("\nFirst 5 rows:\n", df.head())
```

```python
# Check missing values
print("\nMissing Values:\n", df.isnull().sum())

# Unique values in each column
print("\nUnique values per column:\n", df.nunique())
```

```python
# Remove duplicate rows (if any)
df.drop_duplicates(inplace=True)

# Double-check dataset info
df.info()
```

```python
# Create a new column: Age of the startup
current_year = 2025
df['Startup Age'] = current_year - df['Founded Year']
```

```python
# Confirm categorical columns
categorical_cols = ['Country', 'City', 'Industry', 'Funding Stage']
for col in categorical_cols:
    df[col] = df[col].astype('category')
```

```python
df.drop(columns=['Startup Name'], inplace=True)
```

```python
# Selecting numeric and categorical features
features = ['Country', 'City', 'Industry', 'Funding Stage', 'Total Funding (USD)', 
            'Valuation (USD)', 'Revenue (USD)', 'Number of Employees', 'Startup Age']

target = 'Success'

X = df[features]
y = df[target]
```

```python
from sklearn.preprocessing import LabelEncoder

X_encoded = X.copy()
label_encoders = {}

# Encode categorical columns
for col in X.select_dtypes(include='category').columns:
    le = LabelEncoder()
    X_encoded[col] = le.fit_transform(X[col])
    label_encoders[col] = le
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

feature_importances = pd.Series(model.feature_importances_, index=X_encoded.columns)
feature_importances = feature_importances.sort_values(ascending=True)

plt.figure(figsize=(10, 6))
feature_importances.plot(kind='barh', color='dodgerblue')
plt.title('Feature Importance for Startup Success Prediction')
plt.grid(True)
plt.show()
```
